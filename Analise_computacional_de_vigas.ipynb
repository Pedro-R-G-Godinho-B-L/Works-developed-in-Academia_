```python
import numpy as np
import matplotlib.pyplot as plt

# Comprimento da viga

L = 5.0

# Numero de subintervalos
N = 20

# Pontos da discretizacao
x = np.linspace(0, L, N + 1)

# Espacamento
dx = L / N

print(x)
print('Delta x =', dx)

# Carregamento distribuido [ kN / m]

w = 20.0

# Valor de q em cada ponto da malha


q = (-4*x + 20) * np.ones(N + 1)

# Representacao grafica
plt.plot(x, q, 'o-')
plt.xlabel('x (m)')
plt.ylabel('q (kN/m)')
plt.grid()
plt.show()

# Reacoes de apoio



VA = w * L / 3
VB = w * L / 6

print('Reacoes de apoio :')
print(f'VA = {VA:.2f} kN')
print(f'VB = {VB:.2f} kN')

# Vetor do esforco cortante
Q = np.zeros(N + 1)

# Valor inicial
Q[0] = VA

# Integracao do carregamento
for i in range(N):
    Q[i + 1] = Q[i] - (q[i] + q[i + 1]) / 2 * dx

# Diagrama de esforco cortante
plt.plot(x, Q, 'o-')
plt.axhline(0, color='k')
plt.xlabel('x (m)')
plt.ylabel('Q (kN)')
plt.grid()
plt.show()

# Vetor do momento fletor
M = np.zeros(N + 1)

# Valor inicial
M[0] = 0.0

# Integracao do esforco cortante
for i in range(N):
    M[i + 1] = M[i] + (Q[i] + Q[i + 1]) / 2 * dx

# Diagrama de momento fletor
plt.plot(x, M, 'o-')
plt.axhline(0, color='k')
plt.xlabel('x (m)')
plt.ylabel('M (kN.m)')
plt.grid()
plt.show()

print(f'M(0) = {M[0]:.4f} kN.m')
print(f'M(L) = {M[-1]:.4f} kN.m')

Mmax_num = np.max(M)

Mmax_exato = w * L ** 2 / (9 * np.sqrt(3))

print(f'Mmax numerico = {Mmax_num:.4f} kN.m')
print(f'Mmax exato = {Mmax_exato:.4f} kN.m')
```
